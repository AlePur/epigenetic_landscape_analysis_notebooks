```python
import pickle
with open("../go_reg/reg.pickle", "rb") as f:
    reg = pickle.load(f)
with open("../go_reg/gene-names.pickle", "rb") as f:
    names = pickle.load(f)
```

```python
scit.set_defaults(figsize=(7,6))
```

```python
gene = 'Obp99c'
meth = scit.tl.make_layer_config('meth_imp', in_obsm=True, feature_names=names, feature_active_threshold=0)
scit.tl.layer_pick_features(adata, meth,  gene)
scit.tl.copy_to_obs(adata, meth)
pl = scit.pl.embedding2d(adata, 'X_umap', 'copied_data', cmap = 'Reds', title = f'{gene} methylation',
                    vminmax= (0, np.quantile(adata.obs['copied_data'], q = 0.99)), show = False)
```
